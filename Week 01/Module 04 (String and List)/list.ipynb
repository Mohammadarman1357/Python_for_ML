### List Declaration

```python
## declaration
numbers = [100 , 20, 30 , 40 , 50 , 100 , 100]  # int 
num = [10.20, 3.15, 2.4, 2.8, 3.5]  # float
fruits = ['apple', 'orange', 'lichi']   # string
mix = [10, 10.20, 'apple']     # mixed list 

print ("integer list : ",numbers)
print ("float list : ",num)
print ("String list : ", fruits)
print ("Mixed list :",mix)

list_er_modde_list = [[100, 20, 30], [14, 15 , 16], [13, 12, 12]]

print("List inside list : ",list_er_modde_list)

## access
print("numbers [2] :",numbers[2])

numbers[2] = 90
print("after change numbers[2] : ",numbers[2])

new_list = numbers[0:3]
print("New List : ",new_list)

print(type(num))
```

### List Modifi

```python
## adding a item 
numbers = [10, 20, 30, 40]
print(numbers)

# append 
numbers.append(80)
print(numbers)

# insertion at any position 
numbers.insert(0, 45)  # (position, value)
print(numbers)

# deleting the last value 
numbers.pop()
print(numbers)

# deleting a specific values 
numbers.remove(10)
print(numbers)
```

### List Sort

```python
# sorting 
numbers.sort()
print(numbers)

# reverse sorting
numbers.reverse()
print(numbers)

# sorting descending or reverse
numbers.sort(reverse=True)
print(numbers)
```

### List as Stack

```python
stack = []

# push on stack - insert on stack 
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)

print(stack)
print(f"top element : {stack[-1]}")

# pop into stack - removing the top element 
stack.pop()
print("pop top : ", stack)

# top element 
print(f"top element : {stack[-1]}")
```

### List as Queue

```python
queue = []

# insertion || enqueue
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

print(queue)

# remove || dequeue 
print(queue.pop(0))
print(queue)

# accessing the front element 
print(f"front e {queue[0]}")

queue.pop(0)
print(queue)

print(f"front e {queue[0]}")
```

### List Comprehension

```python
even = []

# naive
for i in range (100) : 
    if i%2 ==0:
        even.append(i)

# print(even)

random_list = [x + 10 for x in range(1,101) if x%2 !=0]
print(random_list)

fruits = ['ApPle', 'OraNGe', 'LicHI']

# lower case
lower_fruits = [fruit.lower() for fruit in fruits]
print(lower_fruits)

# upper case
upper_fruits = [fruit.upper() for fruit in fruits]
print(upper_fruits)
```
