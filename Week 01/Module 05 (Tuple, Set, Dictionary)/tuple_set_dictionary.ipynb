# **Tuple**

### Tuple Declaration

```python
# declaration
tup = (10,20,30,40)

print(type(tup))

float_tup = (10.5,10.2)
mixed_tup = (10, 10.5, "str",True)

lst = [10,20,30]
tup1 = tuple(lst)

print(type(tup1), tup1)
```

### Tuple Access

```python
# access
tup = (10,20,30,10.5,"str",False)

# index
print(tup[3])

# slicing
new_tup = tup[0:3]
print(new_tup)
```

### Mutable and Immutable

```python
# mutable
lst = [10,20,30,40]

lst.append(60)
print(lst)

lst[1] = 100
print(lst)

# immutable
tup = (10,20,30,40)
# tup[1] = 100
# print(tup)
```

### Tuple Method

```python
tup = (10,20,30,40,40,50,60,10,20)
print(tup.count(10))
print(tup.index(40))
```

# **Set**

```python
A = {1,2,3}   # set
print(type(A), A)

B = []      # list
C = ()      # tuple
D = {}      # dictionary
E = {1,2}   # set
F = {"name" : "Arman"}  # dictionary

print(type(B))
print(type(C))
print(type(D))
print(type(E))
print(type(F))
```

```python
S = set()
print(type(S))
```

### Access Set

```python
S = {1,2,3,4,1,2}

if(3 in S ):
  print("3 is present")
else:
  print("3 is not present")

sum = 0

for element in S :
  print(element)
  sum+=element  # sum 1 + 2 + 3 + 4

print(sum)
```

### Mutable and Immutable

```python
# mutable
s = {5,1,2,4,3}
print(s)

s.add(10)
print(s)

s.add(0)
print(s)

s.pop()
s.pop()     # pop from front
print(s)

s.remove(5) # remove specific value
print(s)
```

### Set Mathematical Methods

```python
a = {1,2,3}
b = {1,2,5,4,6}

# union
print(a.union(b))

# intersection
print(a.intersection(b))

# isdisjoint
print(a.isdisjoint(b))

# issubset
print(a.issubset(b))
```

# **Dictionaries in Python**

```python
# empty
dic = {}
print(type(dic))

# dic = {key : value}
dic = {
    "name" : "Phitron",
    "age" : 23,
    "address" : "Dhaka",
    "numbers" : [10,20,30,40]
    }
print(type(dic), dic)
```

```python
# access
# dic_name{key}
print(dic["numbers"])
print(dic.get("age"))

dic["name"] = "Phitron AI/ML"

print(dic)

# age override
dic = {
    "name" : "Phitron",
    "age" : 23,
    "address" : "Dhaka",
    "numbers" : [10,20,30,40],
    "age" : 30
    }
# printing default values
print( dic)

print(dic.get("math_marks"))
print(dic.get("math_marks",0))
```

```python
# adding a value
dic = {
    "name" : "Phitron",
    "age" : 23,
    "address" : "Dhaka",
    "numbers" : [10,20,30,40]
    }

print(dic)

# adding a value
dic["math_marks"] = 30
print(dic)

# update
dic.update({"english_marks" : 40})
print(dic)
```

```python
# deleting
del dic["math_marks"]
print(dic)
```

```python
# copy dictionary
dic_2 = dic.copy()
print(dic_2)
```

### Iterating over Dictionary

```python
dic = {
    "name" : "Phitron",
    "age" : 23,
    "address" : "Dhaka",
    "numbers" : [10,20,30,40],
    "age" : 30
    }
# dic = {keys : values}

# keys
keys = dic.keys()
print(keys)

dic["math_marks"] = 30
print(keys)

# values
values = dic.values()
print(values)

dic["math_marks"] = 40
print(values)

items = dic.items()
print(items)
```

```python
# @title
# iteration
for key, value in dic.items():
  print(key,value)
```

### Dictionary Comprehension

```python
square = {x:x**2 for x in range (1,11) if x%2==0}

print(square)
```

```python
co_ordinates = [(10,10.55),(20.5,12),(101,102)]
locations = ["dhaka","chattogram","sylhet"]

exact_location = {co_or:loc for co_or, loc in zip(co_ordinates, locations)}
print(exact_location)
```

### Given a string , print the frequency of the words

```python
string = """data science machine learning data analysis machine
learning statistics data models data training data validation features
features labels preprocessing data augmentation models data optimization
gradient descent neural networks data tensors matrices visualization
exploration pandas numpy matplotlib seaborn scikitlearn tensorflow pytorch
deployment inference production monitoring reproducibility experiments results
metrics accuracy precision recall f1 cross validation data machine
"""
words = string.split()
count = {}

for word in words :
  count [word] = count.get(word, 0) + 1

for k,v in count.items():
  print(f"count of {k} is {v}")
```
