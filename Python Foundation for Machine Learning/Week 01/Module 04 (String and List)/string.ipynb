### String Methods

```python
string = "Welcome to Phitron ML Course, Phitron ML, Hello ML"

processed_string = string.lower()

# length
print(len(string))

# word ase kina check --> searching
print("phitron" in string)
print("phitron" in processed_string)

# shuror kon index teke substring ta soro hoice 
start_index = processed_string.find("phitron")
print(start_index)

# sheser dike kon index teke substring ta shuru hoice 
shes_index = processed_string.find("phitron")
print(shes_index)

# substring koibar ace 
count = processed_string.count("phitron")
print(count)

# replace a substring 
processed_string = string.replace("ML", "AI/ML")
print(processed_string)
```

### Split and Join String

```python
prompt = "What is Phitron ?"

# split 
tokens = prompt.split()
print(type(tokens))
print(tokens)

# join 
sentence = "-".join(tokens)
print(type(sentence))
print(sentence)
```
