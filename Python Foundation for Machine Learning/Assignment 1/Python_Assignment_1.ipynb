# Question No-1
**Given a list of integers, extract all numbers that are divisible by 4 but not divisible by 8.**

Input Format:
Space-separated integers

Output Format:
A list of numbers satisfying the condition

Example:
Input:
4 6 8 12 16 20 24 28

Output:
[4, 12, 20, 28]


###**Answer to the question no - 1 :**

```python
inp = input()

x = inp.split()

x = [int(n) for n in x]

numbers = []

for i in range(len(x)):
    if x[i] % 4 == 0 and x[i] % 8 != 0:
      numbers.append(x[i])

print(numbers)

```

# Question No - 2

Write a function compute_ratio(x, y) that returns (x + y) / (x * y). Use exception handling to manage division by zero.

**Use exception handling to manage:**
 - Division by zero

**Input Format:**

Two values

**Output Format:**

Computed result OR error message

**Example:**

**Input:**

5 0

**Output:**

Error: Division not possible



###**Answer to the question no - 2**

```python
inp = input()

numbers = inp.split()

x = int(numbers[0])
y = int(numbers[1])


def compute_ratio(x, y):
    try:
        result = (x + y) / (x * y)
        return result
    except ZeroDivisionError:
        return "Error: Division not possible"

print(compute_ratio(x, y))
```

# Question No - 3

Write a Python program that reads a text file and finds the longest word.

**Requirements:**

Open and read data from a file (Example: data.txt)
If multiple words have the same longest length, print any one of them


**Input Format:**

A text file containing multiple lines of text

**Output Format:**

The longest word

**Example File Content (data.txt):**

Python is a wonderful language for machine learning.

**Output:**

wonderful


### **Answer to the question no - 3**

```python
with open("./sample_data/data.txt", "r") as file:
    words = file.read()
    words = words.split()

longest_word = ""

for word in words:
    if len(word) > len(longest_word):
        longest_word = word

print(longest_word)
```

# Question No - 4

Write a program that converts a sentence into a list of words but only includes words that start with a vowel (a, e, i, o, u), ignoring case.

**Input Format:**

A single line sentence

**OutputFormat:**

A list of strings

Example:

Input:

An apple a day keeps the doctor away

Output:

['An', 'apple', 'a', 'away']


### **Answer to the question no - 4**

```python
inp = input()
sentence = inp.split()

vowels = ['a', 'e', 'i', 'o', 'u']

vowel_words = []

for word in sentence:
  if word[0].lower() in vowels:
    vowel_words.append(word)

print(vowel_words)
```

# Question No - 5

A word is called a Palindrome if reversing its letters results in the same word (ignoring case). Write a program to check whether a given string is a palindrome.

**Input Format:**

A single word string

**Output Format:**

"Palindrome" or "Not Palindrome"

Example:

**Input:**

Radar

**Output:**

Palindrome


### **Answer to the question no - 5**

```python
inp = str(input())

lower_str = inp.lower()

for i in range(len(lower_str)):

  if lower_str[i] != lower_str[-i-1]:
    print('Not Palindrome')
    break
else:
    print("Palindrome")

```
