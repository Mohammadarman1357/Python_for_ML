## Transforming to df from inbuilt datas

### list

```python
# list 
my_list = [['Alice',25],['Bob', 30], ['Charlie', 28]]

simple_list_df = pd.DataFrame(my_list)  
list_df = pd.DataFrame(my_list,columns=['Name',"Age"], index=[1,2,3])   # column name set korci , index o nij teke caile dite parbo

print(type(list_df))
list_df
```

### tuple

```python
# tuple 
my_tuple = (('Alice', 25),('Bob',30),('Charlie', 28))

tuple_df = pd.DataFrame(my_tuple,columns=['Name','Age'],index=[1,2,3])

print(type(tuple_df))
tuple_df
```

### dictionary

```python
# dictionary

my_dict = {
    'Name' : ['Alice', 'Bob', 'Charlie'],
    'Age' : [25,30,28]
}

dict_df = pd.DataFrame(my_dict)

print(type(dict_df))
dict_df
```

```python
my_data = [
    {'Name' : 'Alice', 'Age' : 25, 'City': 'New York'},
    {'Name' : 'Bob', 'Age' : 30, 'City': 'Paris'},
    {'Name' : 'Alice', 'Age' : 28}    
]

my_data_df = pd.DataFrame(my_data)

print(type(my_data_df))

my_data_df
```

## Filtering Data based on condition

### condition -- df (not started)

```python
# not started --> condition 
not_started = df.loc[df['Completion Status'] == 'Not Started']

not_started
```

### condition -- df (completed)

```python
# completed --> condition

completed = df.loc[df['Completion Status'] == 'Completed']

completed
```

### multiple condition

```python
# completed and data structure marks is greater than or equal to 90

completed_ds90 = df.loc[(df['Completion Status'] == 'Completed') & (df['Data Structure Marks'] >= 90)]

completed_ds90
```
