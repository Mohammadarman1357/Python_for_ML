## Ndarray

```python
import numpy as np

# 1D Array
arr1 = np.array([1, 2, 3,4,5])
print(arr1)

# 2D Array
arr2 = np.array([ [1, 2, 3],  # row1
                  [4, 5, 6]   # row2
                ])
print(arr2)

# 3D Array
arr3 = np.array([

      # 1st floor
      [[1, 2, 3],  # row1
      [4, 5, 6]],   # row2

      # 2nd floor
      [[1, 2, 3],  # row1
      [4, 5, 6]],   # row2

      # 3rd floor
      [[1, 2, 3],  # row1
      [4, 5, 6]],   # row2
])

print(arr3)
```

## Array attributes

```python
# dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

# shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

# data type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

# size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

## ndarray data type

```python
arr = np.array([1, 2, 3])
print(arr.dtype)

arr = np.array([1.5, 2.4, 3.2])
print(arr.dtype)
```

```python
arr = np.array([1,2,3.2])
# upcasted to float
print(arr.dtype)

# upcasted to string
arr = np.array([1,2,3.2, 'hello'])
print(arr.dtype)

arr = np.array([1, True, 3.25, 'hello'])
print(arr.dtype)
```

## Selecting a data type for an array

```python
arr = np.array([1,2,300], dtype=np.int16) # upcasted to int16
print(arr.dtype)
print(arr)

arr = arr.astype(np.int32) # upcasted to int 32
print(arr.dtype)

# error
try:
  arr = np.array([1,2,3.2, 'hello'], dtype = np.float64)
  print(arr)
except Exception as e:
  print(e)

```

## ndarray creation from existing data (Bonus)

```python
# list

lst = [10, 20, 30, 40, 50]

arr = np.array(lst, dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype)  #int32

# mixed list
mixed_lst = [10, True, 'Hello']

arr = np.array(mixed_lst)
print(arr)
print(type(arr))
print(arr.dtype)  #<U21

matrix = [[1,2,3], #row1
          [4,5,6]  #row2
          ]

arr = np.array(matrix)
print(arr)
print(type(arr))
print(arr.dtype)  #int64
```

```python
#tuple
tpl = (10, 20, 30, 40, 50)

arr = np.array(tpl, dtype=np.int32)
print(arr.dtype)

arr = arr.astype(np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# set

st = {1,2,3}

arr = np.array(list(st))
print(arr.dtype)
arr = arr.astype(np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# dictionary

dc = {'a':1, 'b':2, 'c':3}

keys = dc.keys()
values = dc.values()
items = dc.items()

print(keys)
print(values)
print(items)
```

```python
#keys to ndarray
arr = np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#values to ndarray
arr = np.array(list(values))
print(arr)
arr = arr.astype(np.int16)

print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# items to ndarray
arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)  # dtype --> data type
print(arr.shape)
print(arr.ndim) # ndim --> n dimension
print(arr.size)
```

## Creating ndarray from scratch

```python
# np.zeros   --> value will zero

arr = np.zeros((2,3), dtype=np.int8)

# print(arr)
# print(type(arr))
# print(arr.dtype)
# print(arr.shape)
# print(arr.ndim)
# print(arr.size)

arr = np.zeros_like(arr3) ## arr3 shape same
print(arr)
print(arr.shape)
```

```python
# np.ones --> all values will one

arr = np.ones((3,4,3),dtype=np.int8)
# print(arr)
# print(type(arr))
# print(arr.dtype)
# print(arr.shape)
# print(arr.ndim)
# print(arr.size)

arr = np.ones_like(arr3)  ## arr3 shape same
print(arr)
print(arr.shape)
```

```python
# np.empty

arr = np.empty((2,3), dtype=np.int8)

# print(arr)
# print(type(arr))
# print(arr.dtype)
# print(arr.shape)
# print(arr.ndim)
# print(arr.size)

arr = np.empty_like(arr3)  ## arr3 shape same
print(arr)
print(arr.shape)
```

```python
# np.full

arr = np.full((4,3), np.inf)

# print(arr)
# print(type(arr))
# print(arr.dtype)
# print(arr.shape)
# print(arr.ndim)
# print(arr.size)

arr = np.full_like(arr3, np.inf, dtype=np.float64)  ## arr3 shape same
print(arr)
print(arr.shape)
# print(arr.dtype)
```

## array creation with random values

```python
# np.random.rand(shape)

arr = np.random.rand(2,3) # 0 -> 1

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.radint(start, end, shape)

arr = np.random.randint(1,10,(2,3))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.uniform(start,end,shape)

arr = np.random.uniform(50, 100,(3,3,4))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

## array creation with range functions

```python
# np.arange(first,end,step) --> similar to range function

arr = np.arange(1,10,1)
mat = arr.reshape(3,3)

print(arr)
print(mat)  # matrix
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.linspace(start, end(included), koita)

arr = np.linspace(0,4,6)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.logspace(start, end(included),koita)

arr = np.logspace(0,4,6, base=2)

print(arr)
```

## indexing and slicing

```python
# indexing

print(arr1)
print(arr2)

arr1[2] = 100
print(arr1)

arr2[0][2] = 100
print(arr2)
```

```python
# slicing
# 1d = arr[start : end : step]

arr1_mod = arr1[1:4].copy()
arr1[1] = 2

arr1_mod[2] = 200
print(arr1_mod)
print(arr1)
```

```python
# 2d slicing
# arr[row_start : row_end : step, column_start : column_end : step]

# print(arr2)

# getting a row

row_0 = arr2[0:1,]
row_1 = arr2[1: , ]

print(row_0)
print(row_1)

# getting a column

col_0 = arr2[ ::, 0:1]
col_1 = arr2[ :: , 1:2]

print(col_0)
print(col_1)

# getting a portion

portion = arr2[:: , 0:2]
print(portion)
```
