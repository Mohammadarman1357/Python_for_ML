### Q4.
Create a new NumPy array where each employee’s age is adjusted based on:

Add **4 years** if age is less than the average age

Add **2 years** if age is greater than or equal to the average age

Do NOT modify the original array.

Print:

original array

updated array

average age used for condition


```python
# orginial array
org_arr = np.array(emp_ages)
print(org_arr)

# copy array to updated array
updated_arr = org_arr.copy()

# average age
avg_age = np.average(updated_arr)

# find out the conditions
less_than_age = (org_arr < avg_age)
greater_than_age = (org_arr >= avg_age)

# update the array
updated_arr[less_than_age] += 4
updated_arr[greater_than_age] += 2

print(updated_arr)

print(avg_age)
```

### Q5.
Using NumPy:

Assume all salaries of Tech **roles** increase by 8%

Calculate the **new total salary expense**

Find the **overall increase in expense** after the raise


**Print both old and new totals along with the difference.**


```python
# roles
emp_roles = np.array(roles_list)

role = len(emp_income) // len(emp_roles)

# roles repeat korce. for adjusting the shape of emp_income
emp_roles = np.tile(emp_roles, role)

# salary
salary = emp_income.copy()

increased_salary = salary * 1.08

tech_roles = (emp_roles == "Software Developer") | (emp_roles == "Backend Developer")

increased_salary_by_tech = np.where(tech_roles, increased_salary, salary)

old_salary = np.sum(salary)
new_salary = np.sum(increased_salary_by_tech)

new_total_salary = new_salary - old_salary

print(old_salary)
print(new_salary)
print(new_total_salary)
```

### Q10.
Using .loc:

Increase salary by **12% for "Software Developer" and "Backend Developer"**

Increase salary by **5% for all other roles**

Then:

Print the dataframe that includes **role , previous_salary ,**
**new_salary , increase_of_salary**



```python
# error solving
df['emp_salary'] = df['emp_salary'].astype(float)

# previous salary copy
previous_salary = df['emp_salary'].copy()

# find by tech and non tech
df.loc[(df['role'] == 'Software Developer') | (df['role'] == 'Backend Developer'), 'emp_salary'] *= 1.12
df.loc[(df['role'] != 'Software Developer') & (df['role'] != 'Backend Developer'), 'emp_salary'] *= 1.05

new_salary = df['emp_salary']
increase_of_salary = new_salary - previous_salary

dataframe = pd.DataFrame({
    'role': df['role'],
    'previous_salary': previous_salary,
    'new_salary': new_salary,
    'increase_of_salary': increase_of_salary
})

dataframe
```
