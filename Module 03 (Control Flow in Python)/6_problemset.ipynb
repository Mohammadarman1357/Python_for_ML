### K - Min and Max (Problem Set)

```python
# Given 3 numbers A, B and C, Print the minimum and the maximum numbers.
# Input : Only one line containing 3 numbers A, B and C ( - 105 ≤ A, B, C ≤ 105)
# input -->  1 2 3 || -1 -2 -3
# output --> 1 3   ||  -3 -1

inp = input()

numbers = inp.split()

x = int(numbers[0])
y = int(numbers[1])
z = int(numbers[2])

min = x
max = x

## min
if y < min:
    min = y
if z < min:
    min = z

## max
if y > max:
    max = y
if z > max:
    max = z

print(min, max)
```

### C. Even, Odd, Positive and Negative (Problem Set)

```python
# Given N numbers. Count how many of these values are even, odd, positive and negative.
# input --> -5 0 -3 -4 12
# output :
# Even: 3
# Odd: 2
# Positive: 1
# Negative: 3

n = int(input())  # n - number of input

inp = input()  # input --> 1 2 3 4
numbers = inp.split()

positive = 0
negative = 0
odd = 0
even = 0

for i in range(n):
    x = int(numbers[i])

    # positive or negative
    if x < 0:
        negative += 1
    elif x > 0:
        positive += 1

    # odd or even
    if x % 2 != 0:
        odd += 1
    else:
        even += 1

print("Even:", even)
print("Odd:", odd)
print("Positive:", positive)
print("Negative:", negative)
```

### Q. Digits

```python
# Given a number N. Print the digits of that number from right to left separated by space.
# For each test case print a single line contains the digits of the number separated by space.
# input :           | output :
# 4                 |
# 121               | 1 2 1
# 39                | 9 3
# 123456            | 6 5 4 3 2 1
# 1200              | 0 0 2 1

# test case
t = int(input())

for i in range(t):
    number = int(input())
    n = number

    if n == 0:
        print(0)
        continue

    while n > 0:
        print(n % 10, end=" ")      # end = " " --> next line print na kore space nibe
        n //= 10                    # n = int(n / 10)   same to n//=10
    print()
```
